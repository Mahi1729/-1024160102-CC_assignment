```python
import random

roll_no = int(input("Enter your roll number: "))

random.seed(roll_no)
```

```python
numbers = [random.randint(100, 900) for _ in range(100)]

print(numbers)
```

```python
odd_numbers = [x for x in numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
```

```python
prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

```python
frequency = {}

for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

print(frequency)
```

```python
most_frequent = max(frequency, key=frequency.get)

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", frequency[most_frequent])
```

```python
roll_no = input("Enter your roll number: ")

digits = [int(digit) for digit in roll_no]

print("Digits:", digits)
```

```python
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
```

```python
union = A.union(B)

print("Union:", union)
```

```python
intersection = A.intersection(B)

print("Intersection:", intersection)
```

```python
A_minus_B = A.difference(B)

print("A - B:", A_minus_B)
```

```python
B_minus_A = B.difference(A)

print("B - A:", B_minus_A)
```

```python
symmetric_difference = A.symmetric_difference(B)

print("Symmetric Difference:", symmetric_difference)
```

```python
print("Is A a subset of B?", A.issubset(B))
```

```python
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value X: "))

A.discard(X)

print("Set A after discard:", A)
```
